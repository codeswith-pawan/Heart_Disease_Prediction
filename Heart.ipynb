#**Machine Learning**

```python
X=df.drop('target',axis=1)
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42,stratify=y)
```

#**Feature Engineering**

```python
numerical_features = [
    'age',
    'trestbps',
    'chol',
    'thalach',
    'oldpeak'
]

categorical_features = [
    'sex',
    'cp',
    'fbs',
    'restecg',
    'exang',
    'slope',
    'ca',
    'thal'
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
```

```python
preprocess_data = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)
```

#**Import All the required model and Important query**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score,f1_score,classification_report,confusion_matrix
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import cross_validate
from sklearn.model_selection import StratifiedKFold
```

#**Logistic Regression**

```python
lr_pipeline=Pipeline([
    ('Preprocess',preprocess_data),
    ('model',LogisticRegression(random_state=42))
])
lr_pipeline.fit(X_train,y_train)

Lr_pred=lr_pipeline.predict(X_test)
cm=confusion_matrix(y_test,Lr_pred)
print("Logistic Regression")
print("------------------------------------------------")
print("Accuracy: ",accuracy_score(y_test,Lr_pred),'\n')
print("------------------------------------------------")
print("confusion matrix: \n",cm,"\n")
print("-------------------------------------------------")
print("classification_report:\n",classification_report(y_test,Lr_pred),'\n')
print("-------------------------------------------------")
print('F1_score: ',f1_score(y_test,Lr_pred))
```

#**KNeighborsClassifier**

```python
knn_pipeline=Pipeline([
    ('Preprocess',preprocess_data),
    ('model',KNeighborsClassifier(n_neighbors=5))
])
knn_pipeline.fit(X_train,y_train)

knn_pred=knn_pipeline.predict(X_test)
cm=confusion_matrix(y_test,knn_pred)
print("KNeighborsClassifier")
print("------------------------------------------------")
print("Accuracy: ",accuracy_score(y_test,knn_pred),'\n')
print("------------------------------------------------")
print("confusion matrix: \n",cm,"\n")
print("-------------------------------------------------")
print("classification_report:\n",classification_report(y_test,knn_pred),'\n')
print("-------------------------------------------------")
print('F1_score: ',f1_score(y_test,knn_pred))
```

#**Now We applying Cross Validation**

> After We check is our model performance is incressiong or not


```python
models = {
    'Logistic Regression': LogisticRegression(random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'SVM': SVC(random_state=42),
    'Naive Bayes': GaussianNB(),
    'XGBoost': XGBClassifier(random_state=42),
    'Neural Network': MLPClassifier(random_state=42, max_iter=1000)
}
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
cv_results = []

for name, model in models.items():

    pipeline = Pipeline([
        ('Preprocess', preprocess_data),
        ('model', model)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=['accuracy', 'precision', 'recall', 'f1'],
        n_jobs=-1
    )

    cv_results.append({
        'Model': name,
        'CV Accuracy': scores['test_accuracy'].mean(),
        'CV Precision': scores['test_precision'].mean(),
        'CV Recall': scores['test_recall'].mean(),
        'CV F1': scores['test_f1'].mean()
    })
```

```python
cv_results_df = pd.DataFrame(cv_results)
cv_results_df
```

```python
best_models = {
    'Logistic Regression': lr_grid.best_estimator_,
    'KNN': knn_grid.best_estimator_,
    'Decision Tree': dt_grid.best_estimator_,
    'Random Forest': rf_grid.best_estimator_,
    'SVM': svm_grid.best_estimator_,
    'Naive Bayes': nb_grid.best_estimator_,
    'XGBoost': xgb_grid.best_estimator_,
    'Neural Network': nn_grid.best_estimator_
}
```

```python
from sklearn.metrics import roc_auc_score

roc_results = []

for name, model in best_models.items():

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]

    else:
        y_score = model.decision_function(X_test)
        y_prob = y_score

    auc = roc_auc_score(y_test, y_prob)

    roc_results.append({
        'Model': name,
        'ROC-AUC': auc
    })

roc_results = pd.DataFrame(roc_results)

roc_results
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 7))

for name, model in best_models.items():

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = model.decision_function(X_test)

    fpr, tpr, _ = roc_curve(y_test, y_prob)
    roc_auc = auc(fpr, tpr)

    plt.plot(fpr, tpr, label=f"{name} (AUC = {roc_auc:.3f})")

plt.plot([0, 1], [0, 1], linestyle='--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Model Comparison")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 7))

for name, model in best_models.items():

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = model.decision_function(X_test)

    precision, recall, _ = precision_recall_curve(
        y_test,
        y_prob
    )

    plt.plot(
        recall,
        precision,
        label=name
    )

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve - Model Comparison")
plt.legend()
plt.show()
```
